```python
import pandas as pd

benign = pd.read_csv("BenignTraffic.pcap.csv")
attack = pd.read_csv("DDoS-UDP_Flood.pcap.csv")

print("Benign dataset loaded")
print("Attack dataset loaded")
```

```python
benign_duplicates = benign.duplicated().sum()
attack_duplicates = attack.duplicated().sum()

benign_percentage = benign_duplicates / len(benign) * 100
attack_percentage = attack_duplicates / len(attack) * 100

print("Benign total rows:", len(benign))
print("Benign duplicate percentage:", round(benign_percentage, 2), "%")

print("\nAttack total rows:", len(attack))
print("Attack duplicate percentage:", round(attack_percentage, 2), "%")
```

```python
# Create copies of the original datasets
benign_clean = benign.copy()
attack_clean = attack.copy()

# Remove exact duplicate rows from the copies
benign_clean = benign_clean.drop_duplicates()
attack_clean = attack_clean.drop_duplicates()

# Compare the number of rows
print("Benign rows before:", len(benign))
print("Benign rows after:", len(benign_clean))

print("\nAttack rows before:", len(attack))
print("Attack rows after:", len(attack_clean))
```

```python
# Add labels to each dataset
benign_clean["Label"] = 0
attack_clean["Label"] = 1

# Display a few rows
display(benign_clean.head())
display(attack_clean.head())
```

```python
# Add labels
benign_clean["Label"] = 0
attack_clean["Label"] = 1

# Combine both datasets
data = pd.concat(
    [benign_clean, attack_clean],
    ignore_index=True
)

print("Dataset created successfully!")
print("Shape:", data.shape)
```

```python
print("Combined dataset shape:", data.shape)

print("\nLabel counts:")
print(data["Label"].value_counts())

print("\nLabel percentages:")
print(
    (data["Label"].value_counts(normalize=True) * 100).round(2)
)
```

```python
import numpy as np

# Work on a copy so the original data remains unchanged
data_clean = data.copy()

# Replace infinite values with missing values
data_clean.replace([np.inf, -np.inf], np.nan, inplace=True)

# Count missing values before fixing them
print("Missing values before cleaning:",
      data_clean.isnull().sum().sum())

# Fill missing feature values with each column's median
feature_columns = data_clean.columns.drop("Label")

for column in feature_columns:
    median_value = data_clean[column].median()
    data_clean[column] = data_clean[column].fillna(median_value)

# Check the result
print("Missing values after cleaning:",
      data_clean.isnull().sum().sum())

print("Dataset shape:", data_clean.shape)
```

```python
# Input features: network traffic measurements
X = data_clean.drop("Label", axis=1)

# Target: 0 = benign, 1 = DDoS attack
y = data_clean["Label"]

print("Features shape:", X.shape)
print("Labels shape:", y.shape)

print("\nFirst 5 feature rows:")
display(X.head())

print("\nFirst 5 labels:")
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining label counts:")
print(y_train.value_counts())

print("\nTesting label counts:")
print(y_test.value_counts())
```

```python

import time
import pandas as pd

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    HistGradientBoostingClassifier
)
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Define the models
models = {
    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "Logistic Regression": make_pipeline(
        StandardScaler(),
        LogisticRegression(
            max_iter=300,
            random_state=42
        )
    ),

    "Hist Gradient Boosting": HistGradientBoostingClassifier(
        max_iter=100,
        random_state=42
    )
}

# Store model results
results = []
trained_models = {}

# Train and evaluate each model
for name, model in models.items():
    print(f"\nTraining {name}...")
    start_time = time.time()

    model.fit(X_train, y_train)

    # Predict class labels
    predictions = model.predict(X_test)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(
        y_test, predictions, zero_division=0
    )
    recall = recall_score(
        y_test, predictions, zero_division=0
    )
    f1 = f1_score(
        y_test, predictions, zero_division=0
    )

    # Calculate ROC-AUC using predicted probabilities
    probabilities = model.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, probabilities)

    elapsed = time.time() - start_time

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": auc,
        "Training + Evaluation Seconds": round(elapsed, 2)
    })

    trained_models[name] = model

    print(f"{name} completed in {elapsed:.1f} seconds")

# Display results, best F1 score first
results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    by="F1 Score", ascending=False
).reset_index(drop=True)

print("\nMODEL COMPARISON")
display(results_df.round(4))
```

```python

from sklearn.metrics import confusion_matrix, classification_report
from sklearn.tree import DecisionTreeClassifier

# 1. Examine Random Forest predictions
rf_model = trained_models["Random Forest"]
rf_predictions = rf_model.predict(X_test)

print("RANDOM FOREST CONFUSION MATRIX")
print(confusion_matrix(y_test, rf_predictions))

print("\nClassification Report")
print(classification_report(
    y_test,
    rf_predictions,
    target_names=["Benign", "DDoS Attack"]
))

# 2. Find the most important Random Forest features
importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False)

print("\nTOP 15 FEATURES")
display(importance_df.head(15).round(5))

# 3. Diagnostic: use Protocol Type alone
protocol_model = DecisionTreeClassifier(random_state=42)

protocol_model.fit(
    X_train[["Protocol Type"]],
    y_train
)

protocol_predictions = protocol_model.predict(
    X_test[["Protocol Type"]]
)

print("\nPROTOCOL TYPE ONLY")
print("Accuracy:",
      round(accuracy_score(y_test, protocol_predictions), 4))

print("Confusion matrix:")
print(confusion_matrix(y_test, protocol_predictions))
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Start with the combined data before median imputation
X_raw = data.drop(columns=["Label"]).copy()
y_raw = data["Label"].copy()

# Replace infinite values with missing values
X_raw = X_raw.replace([np.inf, -np.inf], np.nan)

# Split before learning imputation values
X_train_raw, X_test_raw, y_train_raw, y_test_raw = train_test_split(
    X_raw,
    y_raw,
    test_size=0.20,
    random_state=42,
    stratify=y_raw
)

# Learn medians ONLY from training data
imputer = SimpleImputer(strategy="median")

X_train_fixed = pd.DataFrame(
    imputer.fit_transform(X_train_raw),
    columns=X_raw.columns,
    index=X_train_raw.index
)

X_test_fixed = pd.DataFrame(
    imputer.transform(X_test_raw),
    columns=X_raw.columns,
    index=X_test_raw.index
)

# Train a fresh model
rf_fixed = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_fixed.fit(X_train_fixed, y_train_raw)
fixed_predictions = rf_fixed.predict(X_test_fixed)

print("Accuracy:", round(
    accuracy_score(y_test_raw, fixed_predictions), 4
))

print("\nConfusion matrix:")
print(confusion_matrix(y_test_raw, fixed_predictions))

print("\nClassification report:")
print(classification_report(
    y_test_raw,
    fixed_predictions,
    target_names=["Benign", "DDoS-UDP"],
    zero_division=0
))
```

```python
import numpy as np
import pandas as pd

def check_data_quality(df, name):
    numeric_data = df.select_dtypes(include="number")

    missing = numeric_data.isna().sum().sum()
    infinite = np.isinf(numeric_data).sum().sum()

    print(name)
    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Missing values:", missing)
    print("Infinite values:", infinite)
    print("-" * 35)

check_data_quality(benign, "Benign dataset")
check_data_quality(attack, "UDP attack dataset")
check_data_quality(tcp_attack, "TCP dataset")
```

```python
def show_invalid_columns(df, name):
    numeric_data = df.select_dtypes(include="number")

    missing = numeric_data.isna().sum()
    infinite = np.isinf(numeric_data).sum()

    report = pd.DataFrame({
        "Missing Values": missing,
        "Infinite Values": infinite
    })

    report = report[
        (report["Missing Values"] > 0) |
        (report["Infinite Values"] > 0)
    ]

    print("\n", name)
    display(report)

show_invalid_columns(benign, "Benign dataset")
show_invalid_columns(attack, "UDP attack dataset")
show_invalid_columns(tcp_attack, "TCP dataset")
```
